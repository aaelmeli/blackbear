```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

area = 2.00e-4

# Load data from CSV file
df = pd.read_csv('RCBeam_WithSlipNoPlasticity_out.csv')

# Extract the columns (replace 'Column1' and 'Column2' with your actual column names)
x = df['conc_disp_x']
y = df['conc_fx'] + df['bar_fx']/area

# Check for NaN or infinite values and remove them
mask = np.isfinite(x) & np.isfinite(y)
x = x[mask]
y = y[mask]

# Identify the breakpoint where the slope changes
def find_breakpoint(x, y):
    for i in range(1, len(x)-1):
        try:
            slope1, _ = np.polyfit(x[:i], np.abs(y[:i]), 1)
            slope2, _ = np.polyfit(x[i:], np.abs(y[i:]), 1)
            if np.abs(slope1 - slope2) > 1e-6:  # Adjust threshold as needed
                return i
        except np.linalg.LinAlgError:
            continue
    return len(x) // 2  # Fallback to midpoint if no clear breakpoint is found

breakpoint = find_breakpoint(x, np.abs(y))

# Compute the slopes for the two segments
slope1, intercept1 = np.polyfit(x[:breakpoint], np.abs(y[:breakpoint]), 1)
slope2, intercept2 = np.polyfit(x[breakpoint:], np.abs(y[breakpoint:]), 1)

# Plot the data points
plt.plot(x, np.abs(y), marker='o', linestyle='')

# Plot the regression lines for each segment
plt.plot(x[:breakpoint], slope1 * x[:breakpoint] + intercept1, color='red', linestyle='--', label=f'Slope 1: {slope1:.2e}')
plt.plot(x[breakpoint:], slope2 * x[breakpoint:] + intercept2, color='blue', linestyle='--', label=f'Slope 2: {slope2:.2e}')

# Adding labels and title
plt.xlabel('conc_disp_x')
plt.ylabel('conc_fx')
plt.title('Plot of conc_disp_x vs conc_fx')

# Annotate the slopes on the figure in scientific notation
plt.text(x[breakpoint//2], slope1 * x[breakpoint//2] + intercept1, f'Slope 1: {slope1:.2e}', fontsize=8, color='red')
plt.text(x[breakpoint + (len(x) - breakpoint)//2], slope2 * x[breakpoint + (len(x) - breakpoint)//2] + intercept2, f'Slope 2: {slope2:.2e}', fontsize=8, color='blue')

# Plot and annotate the breakpoint
breakpoint_x = x[breakpoint]
breakpoint_y = np.abs(y[breakpoint])
plt.plot(breakpoint_x, breakpoint_y, 'go')  # green point
plt.annotate(f'Breakpoint\n({breakpoint_x:.2e}, {breakpoint_y:.2e})', 
             xy=(breakpoint_x, breakpoint_y), 
             xytext=(breakpoint_x + (x.max() - x.min()) * 0.05, breakpoint_y), 
             arrowprops=dict(facecolor='black', arrowstyle='->'))

# Add grid
plt.grid(True)

# Add legend
plt.legend()

# Show the plot
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Define the parameters
sigma_max = 100000.0  # You can set this to any value you prefer
Delta_1 = 0.03  # You can set this to any value you prefer

# Define the relationship between sigma_s and Delta
def sigma_s(Delta, sigma_max, Delta_1):
    if Delta < Delta_1:
        return sigma_max * (0.5 * (Delta / Delta_1) - 4.5 * (Delta / Delta_1)**2 + 1.4 * (Delta / Delta_1)**3)
    else:
        return 1.9 * sigma_max

# Generate Delta values
Delta_values = np.linspace(0, 2 * Delta_1, 500)

# Compute sigma_s values
sigma_s_values = [sigma_s(Delta, sigma_max, Delta_1) for Delta in Delta_values]

# Plot the curve
plt.plot(Delta_values, sigma_s_values, label=r'$\sigma_s$ vs $\Delta$')
plt.xlabel(r'$\Delta$')
plt.ylabel(r'$\sigma_s$')
plt.title(r'Relationship between $\sigma_s$ and $\Delta$')
plt.legend()
plt.grid(True)

# Show the plot
plt.show()
```
